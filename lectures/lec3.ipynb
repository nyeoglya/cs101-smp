```python
# 함수와 응용하여 사용
def grade(x):
    if x>95:
        return "A+"
    elif 95 >= x > 90:
        return "A0"
    elif 90 >= x > 85:
        return "A-"
    else:
        return "Under B+"

print(grade(100))
print(grade(97))
print(grade(82))
print(grade(93))
print(grade(87.7))
```

```python
# 비어있는 객체는 항상 False, 안 비어 있으면 True
if not []:
    print("not [] == True")

if not ():
    print("not () == True")

if not {}:
    print("not {} == True")

if not dict():
    print("not dict() == True")

if not 0:
    print("not 0 == True")

if 1:
    print("1 == True")

if -1:
    print("-1 == True")

if [-1]:
    print("[-1] == True")
```

```python
# class 검사
def test(x):
    if isinstance(x, int):
        print(f"x is an integer!!!")
    else:
        print("x is not an integer!!!")

test("hello")
test(112)
test(10.1)
```

```python
# range 함수를 사용
for i in range(1, 10, 3):
    print(f"{i}")
```

```python
# break 사용한 반복 중단
x = [0.1*i for i in range(10)]

for item in x:
    if item<5:
        print(f"{item}<5!!!")
    else:
        print("End loop!!!")
        break
```

```python
# list comprehension - 예제: key를 문자 n, value를 숫자 n^2으로 갖는 dict 만들기
x = [(n, n**2) for n in range (1,10)]
d = dict(x)
print(d)
```

```python
# 다차원 배열과 그 응용 - 구구단 출력
x = [[x*n for x in range(1, 10)] for n in range (1,10)]

def no_newline_print(x):
    print(x, end="")

no_newline_print("    ")
for i in range(1,10):
    no_newline_print(f"{i}  ")
print("\n" + "_"*29)

for li in x:
    print(f"{li[0]}|", end=" ")
    for item in li:
        print(f"{item:2d}", end=" ")
    print("")
```

```python
# 재귀 함수
def fac(n):
    if n > 1:
        return n * fac(n-1)
    else:
        return 1

print(fac(5))
```
