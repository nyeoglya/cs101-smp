```python
# insertion 정렬 예시
import copy
import random

num_list = [random.randint(0, 50) for x in range(10)]
original_list = copy.deepcopy(num_list)
result_list = []

i = len(num_list)
while i > 0:
    min_element = num_list[0]
    for x in num_list:
        if x < min_element:
            min_element = x
    num_list.remove(min_element)
    result_list.append(min_element)
    i -= 1

print(f"기존 리스트: {original_list}")
print(f"정렬된 리스트: {result_list}")

# 리스트가 동일한지 여부를 출력함
def is_same():
    for x, y in zip(result_list, sorted(original_list)):
        if x != y:
            print(f"{x}, {y}가 다름!")
            return False
    return True

if is_same():
    print("리스트가 동일함!")
```

```python
def encrypt(text):
    result_list = []
    for character in text:
        result_list.append(f"{ord(character):05d}")
    
    return "".join(result_list)

def decrypt(encrypted_text):
    if len(encrypted_text) % 5 != 0:
        return "암호화된 텍스트가 아닙니다!!"
    
    result_list = []
    for i in range(len(encrypted_text)//5):
        result_list.append(chr(int(encrypted_text[i*5:i*5+5])))
    
    return "".join(result_list)

print(encrypt("안녕하세요"))
print(decrypt("5050445397546164946450836"))
```

```python
# 더 진화된 암호화 함수
import uuid
import random

file_name = str(uuid.uuid1()) + ".txt"

korean_char = [chr(x) for x in range(44032, 55204+1)]
print(korean_char[-10:-1])

random.shuffle(korean_char)
print(korean_char[-10:-1])

with open(file_name, "w") as code_book:
    for index, char in zip(range(len(korean_char)), korean_char):
        code_book.write(f"{44032+index},{char}\n")
```

```python
def load_code(file_name):
    def convert(data):
        result = data.strip().split(",")
        return (int(result[0]), result[1])

    with open(file_name, "r") as code_book:
        text = code_book.readlines()
        text = [convert(line) for line in text]
    
    num2text = dict(text)
    text2num = {}
    for key, value in num2text.items():
        text2num[value] = f"{key:05d}"
    
    return num2text, text2num
```

```python
num2text, text2num = load_code("7de58727-0c5b-11f0-acde-463500000031.txt")
```

```python
# 한국어만 입력가능하다. 한국어 이외의 값은 무시하고 변환한다.
def encrypt2(text, text2num):
    result_list = []
    for character in text:
        if 44032 <= ord(character) <= 55204:
            result_list.append(text2num[character])
    
    return "".join(result_list)
```

```python
def decrypt2(text, num2text):
    if len(text) % 5 != 0:
        return "올바르게 암호화된 텍스트가 아닙니다!!"
    
    result_list = []
    for i in range(len(text)//5):
        result_list.append(num2text[int(text[i*5:i*5+5])])
    
    return "".join(result_list)
```
