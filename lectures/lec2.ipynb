```python
# 다양한 타입 확인해보기
def getType(*vars):
    for i, var in enumerate(vars):
        print(f"a{i+1} 변수의 타입은: {type(var)}")

a1 = 10.0
a2 = 10.
a3 = True
a4 = []
print(a4)
a5 = "[]"
a6 = ""
a7 = ()
a8 = {3, 4}
a9 = {3: 4}

getType(a1, a2, a3, a4, a5, a6, a7, a8, a9)
```

```python
v1 = 110

print("1. v1의 값은: %d" % (v1))
print("2. v1의 값은: {0:d}, {0:d}".format(110))
print(f"3. v1의 값은: {v1}, {v1}")

print("(%5d)" % v1)
print("(%05d)" % v1)

x = "%d" % 100
print(x)

print("%.0f" % v1)

print("%x" % v1)
print("%o" % v1)

print("%c" % 100)
print("%d" % "abcd")
```
