```python
import time
import gmpy2
from tqdm import tqdm
import random

def random_digit_number(n=100):
    return int(''.join(str(random.randint(0, 9)) for _ in range(n)))

def multiply(x, y): # Karatsuba algorithm
    if x < 10 or y < 10:
        return x * y

    n = max(len(str(x)), len(str(y)))
    m = n // 2

    high_x = x // (10**m)
    low_x = x % (10**m)
    high_y = y // (10**m)
    low_y = y % (10**m)

    z0 = multiply(low_x, low_y)
    z1 = multiply((low_x + high_x), (low_y + high_y))
    z2 = multiply(high_x, high_y)

    return (z2 * 10**(2*m)) + ((z1 - z2 - z0) * 10**m) + z0

list1 = []
list2 = []
list3 = []
def calculate():
    x = random_digit_number()
    y = random_digit_number()
    a = gmpy2.mpz(x)
    b = gmpy2.mpz(y)

    start1 = time.time()
    result1 = multiply(x, y)
    end1 = time.time()

    start2 = time.time()
    result2 = x*y
    end2 = time.time()

    start3 = time.time()
    result3 = a*b
    end3 = time.time()

    list1.append(end1-start1)
    list2.append(end2-start2)
    list3.append(end3-start3)

def average(lst):
    if not lst:
        return 0
    return sum(lst) / len(lst)

for i in tqdm(range(50000)):
    calculate()

print(f"수행 시간 평균 1: {average(list1)*1000:.5f}밀리초")
print(f"수행 시간 평균 2: {average(list2)*1000:.5f}밀리초")
print(f"수행 시간 평균 3: {average(list3)*1000:.5f}밀리초")
```
